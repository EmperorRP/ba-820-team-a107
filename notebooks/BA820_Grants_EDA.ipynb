```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

sns.set_style("whitegrid")
plt.figure(figsize = (10,6))

sns.histplot(grants["estimated_funding"], bins = 30, kde = True)
plt.xlabel('Funding Amount ($)')
```

```python
top_10_agencies = grant_opps['agency_name'].value_counts().head(10)
plt.figure(figsize = (10,6))

sns.barplot(x = top_10_agencies.values, y = top_10_agencies.index)
```

```python
grant_opps['posted_date'] = pd.to_datetime(grant_opps['posted_date'], errors = 'coerce')
grant_opps['current_closing_date_for_applications'] = pd.to_datetime(grant_opps['current_closing_date_for_applications'], errors = 'coerce')
grant_opps['application_window_days'] = (grant_opps['current_closing_date_for_applications'] - grant_opps['posted_date']).dt.days
app_windows = grant_opps[(grant_opps['application_window_days'] >=0) * (grant_opps['application_window_days'] <=365)]['application_window_days']

plt.figure(figsize=(10,6))
sns.histplot(app_windows, bins = 30, kde = True)
```

```python
apps_month_to_month = grant_opps.set_index('posted_date').resample('ME').size()

apps_month_to_month.plot()
```
